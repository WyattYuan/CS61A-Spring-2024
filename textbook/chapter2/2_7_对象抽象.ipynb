# 2.7.2 专用方法

```python
class Account:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance



Account.__bool__ = lambda self: self.balance != 0
acct = Account('John')
bool(acct)
```

```python
acct.balance = 100
bool(acct)
```
