## 2.8.1 测量效率

```python
def fib(n):
        if n == 0:
            return 0
        elif n == 1:
            return 1
        else:
            return fib(n - 2) + fib(n - 1)

def count_frames(f):
        def counted(n):
            counted.open_count += 1
            counted.max_count = max(counted.max_count, counted.open_count)
            result = f(n)
            counted.open_count -= 1
            return result
        counted.open_count = 0
        counted.max_count = 0
        return counted
    # 返回一个新的函数 counted，是在 f 的基础上增加了计数功能的包装器函数
fib = count_frames(fib)
print("fib(19) = " + str(fib(19)))
print("fib.open_count = " + str(fib.open_count))
print("fib.max_count = " + str(fib.max_count))
print("fib(24) = " + str(fib(24)))
print("fib.max_count = " + str(fib.max_count))
```

## 2.8.2 记忆化

```python
def memo(f):
    cache = {}
    def memorized(n):
        if n not in cache:
            cache[n] = f(n)
        return cache[n]
    return memorized
```
