## 函数作为参数传入

```python
def improve(update, close, guess=1):
    iterations = 0
    while not close(guess):
        guess = update(guess)
        iterations += 1
        if iterations > 1100:
            break
    return guess


def newton_update(f, df):
    def update(x):
        return x - f(x) / df(x)

    return update

def approx_eq(x, y, tolerance=1e-15):
    return abs(x - y) < tolerance

def find_root(f, df, guess=1):
    def near_zero(x):
        return approx_eq(f(x), 0)

    return improve(newton_update(f, df), near_zero, guess)


print(find_root(lambda x: x**2 - 4, lambda x: 2*x,2))
```

## 嵌套定义函数
将函数定义放在另一个函数定义内。

- 词法作用域 (Lexical Scope)

名称的解析取决于函数在源代码中定义的位置，而不是被调用的位置。这意味着你可以通过阅读代码的物理结构，直接推导出某个变量（如 a）是从哪个外层函数中获取的。

- 闭包 (Closure)

这种“函数对象 + 它的父环境”的组合被称为闭包。

```python
def average(x, y):
    return (x + y) / 2
def sqrt(a):
    def sqrt_update(x):
        return average(x, a/x) # 这里的a可以访问到sqrt函数的参数a
    
    def sqrt_close(x):
        return approx_eq(x * x, a)
    return improve(sqrt_update, sqrt_close)
```

## 柯里化

```python
def curried_pow(x):
    def h(n):
        return pow(x,n)
    return h

curried_pow(2)(3)
```

## Lambda 表达式

$$\text{lambda } \text{arguments} : \text{expression}$$

```python
# 数据：学生姓名、成绩、年龄
students = [
    {"name": "Alice", "score": 88, "age": 20},
    {"name": "Bob", "score": 95, "age": 19},
    {"name": "Charlie", "score": 95, "age": 22},
    {"name": "David", "score": 82, "age": 21}
]

# 按照成绩（score）升序排序
students.sort(key=lambda s: s["score"], reverse=False)
print(students)

# 如果成绩相同，按照年龄（age）升序排序（多级排序）
students.sort(key=lambda s: (s["score"], s["age"]))
print(lambda s: (s["score"], s["age"]))
print(students)
```
